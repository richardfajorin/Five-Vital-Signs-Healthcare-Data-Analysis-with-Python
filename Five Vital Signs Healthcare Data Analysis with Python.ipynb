```python
import matplotlib.pyplot as plt
heart_rate = df.filter(like="Heart rate reading").mean()

plt.figure(figsize=(10, 5))
plt.plot(heart_rate.index, heart_rate.values, marker="o")
plt.title("Average Heart Rate Across Readings")
plt.xlabel("Reading")
plt.ylabel("Heart Rate (bpm)")
plt.xticks(rotation=45)
plt.show()
```

### Average Axis for Each Heart Row

```python
heart_rate = df.filter(like="Heart rate reading").mean(axis=1)
```

### Systolic average for each row

```python
systolic_bp = df.filter(like="Systolic blood pressure").mean(axis=1)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(heart_rate, systolic_bp)

plt.title("Heart Rate vs Systolic Blood Pressure")
plt.xlabel("Average Heart Rate (bpm)")
plt.ylabel("Average Systolic Blood Pressure (mmHg)")

plt.show()
```

```python
correlation = heart_rate.corr(systolic_bp)
print(correlation)
```

```python
heart_rate_avg = df.filter(like="Heart rate reading").mean(axis=1)
systolic_bp_avg = df.filter(like="Systolic blood pressure").mean(axis=1)
diastolic_bp_avg = df.filter(like="Diastolic blood pressure").mean(axis=1)
oxygen_avg = df.filter(like="Oxygen saturation").mean(axis=1)
temperature_avg = df.filter(like="Temperature reading").mean(axis=1)
```

```python
vitals = pd.DataFrame({
    "Heart Rate": heart_rate_avg,
    "Systolic BP": systolic_bp_avg,
    "Diastolic BP": diastolic_bp_avg,
    "Oxygen Saturation": oxygen_avg,
    "Temperature": temperature_avg
})

vitals.describe()
```

### Correlation of Heart Rate and Systolic BP

```python
correlation = vitals["Heart Rate"].corr(vitals["Systolic BP"])

print("Correlation:", correlation)
```

```python
plt.scatter(vitals["Heart Rate"], vitals["Systolic BP"])

plt.title("Average Heart Rate vs Average Systolic Blood Pressure")
plt.xlabel("Average Heart Rate (bpm)")
plt.ylabel("Average Systolic BP (mmHg)")

plt.show()
```

```python
correlation = vitals["Systolic BP"].corr(vitals["Diastolic BP"])

print("Correlation:", correlation)
```

```python
correlation = df["Subject ID"].corr(systolic_bp)

print("Correlation:", correlation)
```
